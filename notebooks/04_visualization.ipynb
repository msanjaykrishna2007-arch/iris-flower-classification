```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

print("Libraries imported successfully")
```

```python
data = pd.read_csv("../data/iris_cleaned.csv")

print("Dataset loaded successfully")
print("Dataset shape:", data.shape)
```

```python
print("Columns:")
print(data.columns.tolist())

print("\nFirst 5 rows:")
print(data.head())
```

```python
sns.pairplot(
    data,
    vars=[
        "sepal_length",
        "sepal_width",
        "petal_length",
        "petal_width"
    ],
    hue="species"
)

plt.suptitle("Pairwise Relationships Between Iris Measurements", y=1.02)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=data,
    x="species",
    y="sepal_length"
)

plt.title("Sepal Length Distribution by Species")
plt.xlabel("Species")
plt.ylabel("Sepal Length")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=data,
    x="species",
    y="sepal_width"
)

plt.title("Sepal Width Distribution by Species")
plt.xlabel("Species")
plt.ylabel("Sepal Width")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=data,
    x="species",
    y="petal_length"
)

plt.title("Petal Length Distribution by Species")
plt.xlabel("Species")
plt.ylabel("Petal Length")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=data,
    x="species",
    y="petal_width"
)

plt.title("Petal Width Distribution by Species")
plt.xlabel("Species")
plt.ylabel("Petal Width")
plt.show()
```

```python
plt.figure(figsize=(8, 6))

correlation = data[
    [
        "sepal_length",
        "sepal_width",
        "petal_length",
        "petal_width"
    ]
].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Iris Measurements")
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=data,
    x="species"
)

plt.title("Number of Samples by Species")
plt.xlabel("Species")
plt.ylabel("Number of Samples")
plt.show()
```

```python
mean_values = data.groupby("species")[
    ["sepal_length", "sepal_width", "petal_length", "petal_width"]
].mean()

mean_values.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Mean Measurements by Species")
plt.xlabel("Species")
plt.ylabel("Mean Value")
plt.xticks(rotation=0)
plt.legend(title="Measurements")
plt.show()
```

```python
median_values = data.groupby("species")[
    ["sepal_length", "sepal_width", "petal_length", "petal_width"]
].median()

median_values.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Median Measurements by Species")
plt.xlabel("Species")
plt.ylabel("Median Value")
plt.xticks(rotation=0)
plt.legend(title="Measurements")
plt.show()
```

```python
mode_values = data.groupby("species")[
    ["sepal_length", "sepal_width", "petal_length", "petal_width"]
].agg(lambda x: x.mode().iloc[0])

mode_values.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Mode Measurements by Species")
plt.xlabel("Species")
plt.ylabel("Mode Value")
plt.xticks(rotation=0)
plt.legend(title="Measurements")
plt.show()
```

```python
variance_values = data.groupby("species")[
    ["sepal_length", "sepal_width", "petal_length", "petal_width"]
].var()

variance_values.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Variance of Measurements by Species")
plt.xlabel("Species")
plt.ylabel("Variance")
plt.xticks(rotation=0)
plt.legend(title="Measurements")
plt.show()
```

```python
std_values = data.groupby("species")[
    ["sepal_length", "sepal_width", "petal_length", "petal_width"]
].std()

std_values.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Standard Deviation of Measurements by Species")
plt.xlabel("Species")
plt.ylabel("Standard Deviation")
plt.xticks(rotation=0)
plt.legend(title="Measurements")
plt.show()
```

```python
min_values = data.groupby("species")[
    ["sepal_length", "sepal_width", "petal_length", "petal_width"]
].min()

max_values = data.groupby("species")[
    ["sepal_length", "sepal_width", "petal_length", "petal_width"]
].max()

comparison = pd.concat(
    [min_values.add_prefix("Min "), max_values.add_prefix("Max ")],
    axis=1
)

comparison.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Minimum and Maximum Measurements by Species")
plt.xlabel("Species")
plt.ylabel("Measurement Value")
plt.xticks(rotation=0)
plt.legend(title="Statistics")
plt.show()
```


## Visualization Findings

1. The pair plot shows the relationships between the four numerical measurements and helps visualize the separation between the three Iris species.

2. The box plots show the distribution, median, spread, and possible outliers of each measurement for the three species.

3. Petal length and petal width show clear differences between the Iris species.

4. The correlation heatmap shows the strength of relationships between the numerical features.

5. The species distribution graph shows that the three species are almost equally represented in the cleaned dataset.

6. The statistical graphs provide visual comparisons of mean, median, mode, variance, standard deviation, minimum, and maximum values.

## Conclusion

The visualizations provide a clear understanding of the Iris dataset. They show the distribution of measurements, relationships between features, differences between species, class balance, and important statistical characteristics. These visual patterns help support the classification model in the later stage of the project.
