```python
# ============================================================
# IRIS FLOWER CLASSIFICATION - INITIAL EDA
# Member A: Dataset & Initial EDA
# ============================================================

import sys
from pathlib import Path

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------------------------
# 1. Set project root and import data loader
# ------------------------------------------------------------

PROJECT_ROOT = Path.cwd().parent
sys.path.append(str(PROJECT_ROOT))

from src.data_loader import load_iris_data

print("=" * 60)
print("IRIS FLOWER CLASSIFICATION - INITIAL EDA")
print("=" * 60)

# ------------------------------------------------------------
# 2. Load dataset
# ------------------------------------------------------------

iris_df = load_iris_data(PROJECT_ROOT / "data" / "iris.data")

print("\nDataset loaded successfully.")

# ------------------------------------------------------------
# 3. Dataset dimensions
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("1. DATASET DIMENSIONS")
print("=" * 60)

print(f"Number of rows    : {iris_df.shape[0]}")
print(f"Number of columns : {iris_df.shape[1]}")

# ------------------------------------------------------------
# 4. Column names
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("2. COLUMN NAMES")
print("=" * 60)

print(iris_df.columns.tolist())

# ------------------------------------------------------------
# 5. Dataset information
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("3. DATASET INFORMATION")
print("=" * 60)

iris_df.info()

# ------------------------------------------------------------
# 6. First 5 records
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("4. FIRST 5 RECORDS")
print("=" * 60)

display(iris_df.head())

# ------------------------------------------------------------
# 7. Last 5 records
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("5. LAST 5 RECORDS")
print("=" * 60)

display(iris_df.tail())

# ------------------------------------------------------------
# 8. Overall descriptive statistics
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("6. OVERALL DESCRIPTIVE STATISTICS")
print("=" * 60)

display(iris_df.describe())

# ------------------------------------------------------------
# 9. Unique species
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("7. SPECIES / CLASSES")
print("=" * 60)

print("Unique species:")
print(iris_df["species"].unique())

print(f"\nNumber of unique species: {iris_df['species'].nunique()}")

# ------------------------------------------------------------
# 10. Species distribution
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("8. SPECIES DISTRIBUTION")
print("=" * 60)

print(iris_df["species"].value_counts())

# ------------------------------------------------------------
# 11. Missing-value check
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("9. MISSING VALUES")
print("=" * 60)

print("Missing values in each column:")
print(iris_df.isnull().sum())

print(f"\nTotal missing values: {iris_df.isnull().sum().sum()}")

# ------------------------------------------------------------
# 12. Duplicate check
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("10. DUPLICATE CHECK")
print("=" * 60)

duplicate_count = iris_df.duplicated().sum()

print(f"Number of duplicate rows: {duplicate_count}")

# ------------------------------------------------------------
# 13. Data types
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("11. DATA TYPES")
print("=" * 60)

print(iris_df.dtypes)

# ------------------------------------------------------------
# 14. Numeric and categorical columns
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("12. COLUMN TYPES")
print("=" * 60)

numeric_columns = iris_df.select_dtypes(include=np.number).columns.tolist()
categorical_columns = iris_df.select_dtypes(exclude=np.number).columns.tolist()

print("Numeric columns:")
print(numeric_columns)

print("\nCategorical columns:")
print(categorical_columns)

# ------------------------------------------------------------
# 15. Initial observations
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("13. INITIAL OBSERVATIONS")
print("=" * 60)

print("""
- The Iris dataset contains 150 records and 5 columns.
- Four columns represent numerical flower measurements:
  sepal length, sepal width, petal length, and petal width.
- The species column represents the target class.
- There are three Iris species in the dataset.
- Each species contains 50 samples.
- The class distribution is balanced.
- The dataset contains no missing values.
- Duplicate records were checked as part of the initial EDA.
- Overall descriptive statistics were examined for the
  numerical features.
""")

print("\n" + "=" * 60)
print("INITIAL EDA COMPLETED")
print("=" * 60)
```
